```python
def kalkulator_keuntungan_agroindustri():
    print("=== 1. KALKULATOR KEUNTUNGAN & KELAYAKAN PROSESING ===")

    # Input data produksi
    biaya_tetap = float(input("Masukkan Biaya Tetap (misal: penyusutan alat) (Rp): "))
    biaya_variabel_per_unit = float(input("Masukkan Biaya Variabel per unit produk (Rp): "))
    harga_jual_per_unit = float(input("Masukkan Harga Jual produk per unit (Rp): "))
    jumlah_produksi = int(input("Masukkan Jumlah Produksi (unit): "))

    # Perhitungan
    total_biaya = biaya_tetap + (biaya_variabel_per_unit * jumlah_produksi)
    total_penerimaan = harga_jual_per_unit * jumlah_produksi
    keuntungan = total_penerimaan - total_biaya

    # Rasio R/C (Revenue-Cost Ratio) untuk mengukur efisiensi
    rc_ratio = total_penerimaan / total_biaya if total_biaya > 0 else 0

    print("\n--- Hasil Analisis ---")
    print(f"Total Biaya Produksi : Rp {total_biaya:,.2f}")
    print(f"Total Penerimaan     : Rp {total_penerimaan:,.2f}")
    print(f"Keuntungan Bersih    : Rp {keuntungan:,.2f}")
    print(f"R/C Ratio            : {rc_ratio:.2f}")

    if rc_ratio > 1:
        print("Status               : USAHA EFEKTIF / LAYAK (R/C > 1)")
    elif rc_ratio == 1:
        print("Status               : IMPAS / BREAK EVEN (R/C = 1)")
    else:
        print("Status               : RUGI / TIDAK LAYAK (R/C < 1)")


def hitung_eoq():
    print("\n=== 2. MANAJEMEN PERSEDIAAN BAHAN BAKU (EOQ) ===")
    print("Menghitung jumlah pembelian bahan baku pertanian paling optimal.")

    # Input data persediaan
    D = float(input("Kebutuhan bahan baku per tahun (kg/unit): "))
    S = float(input("Biaya pemesanan setiap kali pesan (Rp): "))
    H = float(input("Biaya penyimpanan per unit per tahun (Rp): "))

    if H > 0:
        # Rumus EOQ = sqrt((2 * D * S) / H)
        eoq = ((2 * D * S) / H) ** 0.5
        frekuensi_pesan = D / eoq

        print("\n--- Hasil EOQ ---")
        print(f"Jumlah Pembelian Optimal (EOQ) : {eoq:.2f} kg/unit per pesanan")
        print(f"Frekuensi Pemesanan Ideal      : {frekuensi_pesan:.1f} kali/tahun")
    else:
        print("Biaya penyimpanan harus lebih besar dari 0.")


def main():
    print("=========================================")
    print(" PROGRAM SEDERHANA AGROINDUSTRI ")
    print("=========================================\n")

    kalkulator_keuntungan_agroindustri()
    hitung_eoq()

if __name__ == "__main__":
    main()
```
